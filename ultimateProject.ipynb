````python
'OfficeStackAsyncHaveWorkedWith' in df.columns
````

````python
df['OfficeStackAsyncHaveWorkedWith'].value_counts()
````

````python
# this code splits each and every office stack, and counts the number of thier occurences in the column - `OfficeStackAsyncHaveWorkedWith`
def each_stack(col_index):
    commonOfficeStack = {}
    for i in col_index:
        if len(i) == 1:
            if i not in commonOfficeStack:
                commonOfficeStack[i] = 1
        else:
            a = i.split(';')
            for j in a:
                if j in commonOfficeStack:
                    commonOfficeStack[j] += 1
                else:
                    commonOfficeStack[j] = 1
    return commonOfficeStack
each_stack(df['OfficeStackAsyncHaveWorkedWith'].value_counts().index)
````

````python
data_engineer_grdf = df.groupby(by = 'DevType')
data_engineer_grdf.groups
````

````python
# we need to clean this column as it contains many nan values, just fillna('Nan'), to make the float nan a string nan, `each_stack` function will count this nan as another string object and also count its occurences.
````
